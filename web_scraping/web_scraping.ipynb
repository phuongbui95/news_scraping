```python
import requests
from bs4 import BeautifulSoup
```

```python
# Sample prompt for AI:

'''
Write a Python script using requests and beautifulsoup to collect the information from this article: 
https://vnexpress.net/tieng-viet-nua-tay-nua-ta-5001391.html

Collect the following information:
+ Article's title
+ Article's author
+ Article's content

This is the source code of the article:
<div class="sum-info">
<h3 class="title-news"><a data-medium="Item-NguyenTanDai" href="https://vnexpress.net/tac-gia/nguyen-tan-dai-1832.html" title="Nguyễn Tấn Đại" data-itm-source="#vn_source=Detail-GocNhin_VanHoaLoiSong-5001391&amp;vn_campaign=Box-TacGia&amp;vn_medium=Item-NguyenTanDai&amp;vn_term=Desktop" data-itm-added="1">Nguyễn Tấn Đại</a></h3>
<p class="description">Nhà nghiên cứu độc lập</p>
</div>

'''
```

```python
# Target URL
url = "https://vnexpress.net/tieng-viet-nua-tay-nua-ta-5001391.html"

# Send HTTP GET request
response = requests.get(url)
response.raise_for_status()  # Raise error if request fails

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

# ---------- 1. Article Title ----------
# VnExpress article titles are usually in <h1 class="title-detail">
title_tag = soup.find("h1", class_="title-detail")
title = title_tag.get_text(strip=True) if title_tag else "No title found"

# ---------- 2. Article Author ----------
# From the snippet, the author block is:
# <div class="sum-info"> ... <h3 class="title-news"><a ...>Nguyễn Tấn Đại</a></h3> ...
author = ""
author_div = soup.find("div", class_="sum-info")
if author_div:
    author_link = author_div.find("h3", class_="title-news")
    if author_link and author_link.a:
        author = author_link.a.get_text(strip=True)

if not author:
    author = "No author found"

# ---------- 3. Article Content ----------
# Content paragraphs are usually inside <article> or <section class="fck_detail">
content = ""
content_container = soup.find("article") or soup.find("section", class_="fck_detail")

if content_container:
    paragraphs = content_container.find_all("p")
    content = "\n\n".join(p.get_text(strip=True) for p in paragraphs)
else:
    content = "No content found"

# ---------- Print Results ----------
print("TITLE:")
print(title)
print("\nAUTHOR:")
print(author)
print("\nCONTENT:")
print(content)
```

Write a Python script using requests and beautifulsoup to generate the information from a source code: 

Collect the following information for each item: 

+ Article's title 
+ Article's author 
+ Article's link 

I'm going to show your the source code of item-1. Have a look at it and give me a script to scrap all info of all items

Source code: [paste "Edit as HTML" here]

```python
url = "https://vnexpress.net/goc-nhin"

# Fetch page
response = requests.get(url)
response.raise_for_status()

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

# Container
paging = soup.find("div", id="paging")

articles_data = []

if paging:
    articles = paging.find_all("article", class_="item-news")

    for article in articles:
        # ---------- TITLE + LINK ----------
        title_tag = article.find("h3", class_="title-news")
        title = link = None

        if title_tag and title_tag.a:
            title = title_tag.a.get_text(strip=True)
            link = title_tag.a.get("href")

        # ---------- AUTHOR ----------
        author = None
        author_tag = article.select_one("p.meta-news a.name-author")
        if author_tag:
            author = author_tag.get_text(strip=True)

        articles_data.append({
            "title": title,
            "author": author,
            "link": link
        })

# ---------- OUTPUT ----------
for i, item in enumerate(articles_data, 1):
    print(f"Item {i}")
    print("Title :", item["title"])
    print("Author:", item["author"])
    print("Link  :", item["link"])
    print("-" * 50)
```
